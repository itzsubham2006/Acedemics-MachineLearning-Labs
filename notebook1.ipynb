#### **EXP-2 : Linear Rgression on 200 data points**

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import mean_squared_error, mean_absolute_error 
from sklearn.model_selection import train_test_split
```

```python
x = np.random.rand(200, 1) * 10                        
y = 3 * x.ravel() + 5 + np.random.randn(200) * 2       

```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
print(f"Slope (m) : {round(model.coef_[0])}")
print(f"Intercept (c) : {round(model.intercept_)}")
```

```python
y_pred = model.predict(x_test)
```

```python
print(f"MAE : {mean_absolute_error(y_test, y_pred)}")
```

```python
plt.Figure(figsize=(6, 5))

plt.scatter(x_train, y_train, label = "Train", alpha=0.6)
plt.scatter(x_test, y_test, label = 'Test', alpha=0.8)
xs = np.linspace(0, 10, 100).reshape(-1, 1)

plt.plot(xs, model.predict(xs), label = "Regression Line")

plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```
